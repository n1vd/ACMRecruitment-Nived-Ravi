```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("cleaned_students_performance.csv")

df.head()
```

```python
plt.hist(df["math score"], bins=10)

plt.title("Distribution of Math Scores")
plt.xlabel("Math Score")
plt.ylabel("Number of Students")

plt.show()
```

```python
gender_scores = df.groupby("gender")[["math score", "reading score", "writing score"]].mean()

gender_scores.plot(kind="bar")

plt.title("Average Scores by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Score")
plt.xticks(rotation=0)

plt.show()
```

```python
completed = df[df["test preparation course"] == "completed"]["math score"]
none = df[df["test preparation course"] == "none"]["math score"]

plt.boxplot([completed, none], labels=["Completed", "None"])

plt.title("Math Scores by Test Preparation")
plt.xlabel("Test Preparation")
plt.ylabel("Math Score")

plt.show()
```

```python
plt.scatter(df["reading score"], df["writing score"])

plt.title("Reading Score vs Writing Score")
plt.xlabel("Reading Score")
plt.ylabel("Writing Score")

plt.show()
```

```python
scores = df[["math score", "reading score", "writing score"]]

correlation = scores.corr()

print(correlation)
plt.imshow(correlation)

plt.title("Correlation Between Exam Scores")

plt.xticks([0, 1, 2], ["Math", "Reading", "Writing"])
plt.yticks([0, 1, 2], ["Math", "Reading", "Writing"])

plt.colorbar()

plt.show()
```

```python
## Key Insights

1. Most students have math scores around the middle to higher score range.

2. Female students generally perform better in reading and writing, while male students tend to have a higher average math score.

3. Students who completed the test preparation course generally have better scores than students who did not complete it.

4. Reading and writing scores have a strong positive relationship. Students with high reading scores usually have high writing scores.

5. Math, reading and writing scores are positively correlated with each other.

6. Reading and writing have a stronger relationship with each other than with math scores.
```
